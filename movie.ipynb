```python
#removing the stopwords
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re  
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words=set(stopwords.words('english'))
lemmatizer=WordNetLemmatizer()
```

```python
def preprocess_text(text):
    #lowercase
    text=str(text).lower()
    #removd the punctuations
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    words=text.split()
    #remove stopwords
    words=[word for word in words if word not in stop_words]

    #lemmatize
    words=[lemmatizer.lemmatize(word) for word in words]

    return " ".join(words)
```

```python
df['tags']=df['tags'].apply(preprocess_text)
```

```python
df['tags'][3]
```

```python
#removing the missingg index
df=df.reset_index(drop=True)
```

```python
#knowing the index
indices=pd.Series(df.index,index=df['names']).drop_duplicates()
indices
```

```python
#using tfidf 
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf=TfidfVectorizer(max_features=15000,ngram_range=(1,2),stop_words='english')      #ngram is used to vetorize the connected words
```

```python
#creating a matrix that will arrange the format
tfidf_matrix=tfidf.fit_transform(df['tags'])
```

```python
tfidf_matrix 
```

```python
#using cosine similarities
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(names,n=10):
    if names not in indices:
        return['Movie not found']
    
    idx=indices[names]
    sim_score=cosine_similarity(tfidf_matrix[idx],tfidf_matrix).flatten()
    similar_idx=sim_score.argsort()[::-1][1:n+1]
    return df['names'].iloc[similar_idx]
```

```python
recommend('Scooby-Doo',5)
```

```python
#creating pickle files
import pickle

pickle.dump(tfidf_matrix,open('tfidf_matrix.pkl','wb'))

pickle.dump(indices,open('indices.pkl','wb'))

df.to_pickle('df.pkl')

pickle.dump(tfidf,open('tdidf.pkl','wb'))
```
